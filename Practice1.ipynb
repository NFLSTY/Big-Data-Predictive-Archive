```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
print(spark)
print(spark.version)
```

```python
factors = spark.read.csv("/content/drive/MyDrive/Semester 5/Data Predictive Analytics/Week1/StudentPerformanceFactors.csv", header=True, inferSchema=True)
factors.show()
factors
```

**Sorting**

```python
from pyspark.sql.functions import asc, desc

sort_by_exam_score = factors.sort(desc("exam_score"))
sort_by_exam_score.show()
```

**Filtering**

```python
male_study_hours1 = factors.filter("Hours_Studied > 20 AND Gender == 'Male'")
male_study_hours2 = factors.filter((factors.Attendance > 80) & (factors.Gender == "Male"))
```

```python
male_study_hours1.show()
```

```python
male_study_hours2.show()
```

**Selecting**

```python
selected1 = factors.select('Parental_Involvement','Sleep_Hours', 'Tutoring_Sessions', 'School_Type', 'Teacher_Quality', 'Exam_Score')
temp = factors.select(factors.Parental_Involvement, factors.Sleep_Hours, factors.Tutoring_Sessions, factors.School_Type, factors.Teacher_Quality, factors.Exam_Score)

filterA = factors.Parental_Involvement == "Medium"
filterB = factors.Tutoring_Sessions >= 5
filterC = factors.Teacher_Quality == "High"

selected2 = selected1.filter(filterA).filter(filterB).filter(filterC)
print(selected2.show())
```

```python
selected3 = temp.filter(filterA).filter(filterB).filter(filterC)
print(selected3.show())
```

```python
avg_study_hours = (factors.Exam_Score/factors.Hours_Studied).alias("avg_study_hours")
avg1 = factors.select('Gender', 'Parental_Involvement', 'Motivation_Level', 'Tutoring_Sessions', avg_study_hours)
avg2 = factors.selectExpr('Gender', 'Parental_Involvement', 'Motivation_Level', 'Tutoring_Sessions', "Exam_Score/Hours_Studied AS avg_study_hours")
```

```python
avg1.show()
```

```python
avg1.show()
```

**Aggregating**

```python
factors = factors.withColumn('Hours_Studied', factors.Hours_Studied.cast('int'))
factors = factors.withColumn('Tutoring_Sessions', factors.Tutoring_Sessions.cast('int'))

factors.describe('Hours_Studied', 'Tutoring_Sessions').show()
```

```python
fc1 = factors.filter(factors.Motivation_Level == "High").groupBy().min("Hours_Studied")
fc2 = factors.filter(factors.Internet_Access == "Yes").groupBy().max("Sleep_Hours")
```

```python
fc1.show()
```

```python
fc2.show()
```

```python
flg3 = factors.filter(factors.Motivation_Level == "High").filter(factors.Internet_Access == "Yes").groupBy().avg("Hours_Studied")
flg3.show()
```
